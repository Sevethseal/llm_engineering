```python
MODEL = 'gpt-4o-mini'
openai = OpenAI()

user_prompt="Explian bubbling in javascript"
system_prompt="""
You are a helpful assistant. For every response, format using:
- A main title with `#`
- Section headings with `##` and subheadings with `###`
- Numbered lists for processes
- Bullet lists for key points
- **Bold** for emphasis
- `inline code` and fenced code blocks for code examples
- Horizontal rules (`---`) to separate major sections
"""
```

```python
def ask_and_render(question: str, model: str = "gpt-4o-mini"):
    """
    Send a technical question to GPT and display the answer as styled Markdown in Jupyter.
    
    Parameters:
    - question: The user’s technical question.
    - model:     The OpenAI model to use (default: "gpt-4o-mini").
    """
    response = openai.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user",   "content": question}
        ]
    )
    markdown_answer = response.choices[0].message.content
    display(Markdown(markdown_answer))
```

```python
ask_and_render("How to use chat gpt")
```
